```python
#load and plot the downloed images 
from matplotlib import pyplot as plt

img_color1 = cv2.imread('dog1.jpg')
img_color1 = cv2.cvtColor(img_color1, cv2.COLOR_BGR2RGB)

img_color2 = cv2.imread('dog2.jpg')
img_color2 = cv2.cvtColor(img_color2, cv2.COLOR_BGR2RGB)
img_color3 = cv2.imread('dog3.jpg')
img_color3 = cv2.cvtColor(img_color3, cv2.COLOR_BGR2RGB)
img_color4 = cv2.imread('dog4.jpg')
img_color4 = cv2.cvtColor(img_color4, cv2.COLOR_BGR2RGB)
img_color5 = cv2.imread('panda.jpg')
img_color5 = cv2.cvtColor(img_color5, cv2.COLOR_BGR2RGB)


plt.imshow(img_color1)
plt.show()
plt.imshow(img_color2)
plt.show()
plt.imshow(img_color3)
plt.show()
plt.imshow(img_color4)
plt.show()
plt.imshow(img_color5)
plt.show()
```

```python


import imagehash
from PIL import Image

#------------------- AVERAGE HASH------------------
hash1 = imagehash.average_hash(Image.open('dog1.jpg'))
hash2 = imagehash.average_hash(Image.open('dog2.jpg'))
hash3 = imagehash.average_hash(Image.open('dog3.jpg'))
hash4 = imagehash.average_hash(Image.open('dog4.jpg'))
hash5 = imagehash.average_hash(Image.open('panda.jpg'))

print(hash1)
print(hash2)
print(hash3)
print(hash4)
print(hash5)

print('\n')
print('original - flipped:')
print(hash1 - hash2)

print('original - saturated:')
print(hash1 - hash3)

print('original - cropped:')
print(hash1 - hash4)

print('original - panda:')
print(hash1 - hash5)





```

```python
#------------------- PERCEPTIVE HASH------------------
hash1 = imagehash.phash(Image.open('dog1.jpg'))
hash2 = imagehash.phash(Image.open('dog2.jpg'))
hash3 = imagehash.phash(Image.open('dog3.jpg'))
hash4 = imagehash.phash(Image.open('dog4.jpg'))
hash5 = imagehash.phash(Image.open('panda.jpg'))


print(hash1)
print(hash2)
print(hash3)
print(hash4)
print(hash5)

print('\n\n')
print('original - flipped:')
print(hash1 - hash2)

print('original - saturated:')
print(hash1 - hash3)

print('original - cropped:')
print(hash1 - hash4)

print('original - panda:')
print(hash1 - hash5)
```
